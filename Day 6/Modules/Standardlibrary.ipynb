#### Standard Library Overview
Python's standard library is a vast collection of modules and packages that come bundled with Python, providing a wide range of functionalities out of the box.

```python
import array
arr=array.array('i',[1,2,3,4,5])
print(arr)
```

```python
import math
print(math.sqrt(16))
print(math.pi)
```

```python
## random
import random
print(random.randint(1,10))
print(random.choice(['apple','banana','cherry']))
```

```python
## File and Directory Access
import os
print(os.getcwd())
```

```python
os.mkdir('test_dir')
```

```python
## High level operations on files and collections of files
import shutil
shutil.copyfile('source.txt','destination.txt')
```

```python
## Data Serialization
import json
data = {'name':'Subodh','age':31,'city':'New York'}

json_str = json.dumps(data)
print(json_str)
print(type(json_str))

parsed_data=json.loads(json_str)
print(parsed_data)
print(type(parsed_data))
```

```python
## csv
import csv
with open('example.csv',mode='w',newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['Name', 'Age', 'City'])
    writer.writerow(['Subodh', 31, 'New York'])

with open('example.csv',mode='r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python
## Datetime
from datetime import datetime,timedelta

now = datetime.now()
print("Current date and time:",now)

yesterday = now - timedelta(days=1)
print("Yesterday's date and time:",yesterday)
```

```python
## time
import time
print("Current time:",time.time())

time.sleep(2)
print("Current time after 2 seconds:",time.time())
```

```python
## Regular Expressions
import re
pattern = r'\d+'
text = 'There are 12 apples and 34 bananas.'
match = re.search(pattern,text)
print(match.group())  # Output: 12
```

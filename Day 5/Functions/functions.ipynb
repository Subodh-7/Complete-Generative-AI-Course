#### Introduction to Functions
A function is a block of code that performs a specific task. Functions help in organizing code, resuing code and improving readability.

```python
## syntax
def function_name(parameters):
    """Docstring"""
    # Function body
    return expression
```

```python
## why functions?
num=24
if num%2==0:
    print("The number is even")
else:
    print("The number is odd")
```

```python
def even_or_odd(num):
    """This function finds even or odd"""
    if num%2==0:
        print("The number is even")
    else:
        print("The number is odd")   
```

```python
## Call this function
even_or_odd(24)
```

```python
## function with multiple parameters
def add(a,b):
    return a+b

result=add(2,4)
print(result)
```

```python
## Default Parameter
def greet(name):
    print(f"Hello {name} Welcome to the paradise")

greet("Subodh")
```

```python
def greet(name="Guest"):
    print(f"Hello {name} Welcome to the paradise")

greet()
```

```python
def greet(name="Guest"):
    print(f"Hello {name} Welcome to the paradise")

greet("Subodh")
```

```python
## Variable Length Arguments
## Positional and Keyword arguments
#$ def print_numbers(num1, num2, num3, num4, num5)

#Here we are writing krish but the best practice is to write args
def print_numbers(*krish):
    for number in krish:
        print(number)
```

```python
print_numbers(1,2,3,4,5,6,7,8,"Krish")
```

```python
def print_numbers(*args):
    for number in args:
        print(number)
```

```python
print_numbers(1,2,3,4,5,6,7,8,"Subodh")
```

```python
### Keyword Arguments
def print_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(name="Subodh", age="31", country="India")
```

```python
def print_details(*args,**kwargs):
    for val in args:
            print(f"Positional argument: {val}")

    for key, value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(1,2,3,4,"Krish", name="Subodh", age="31", country="India")
```

```python
## Return Statement
def multiply(a,b):
    return a*b

multiply(2,3)
```

```python
## Return multiple parameters
def multiply(a,b):
    return a*b,a

multiply(2,3)
```

### Introduction to Dictionaries

Dictionaries are unordered collections of items. They store data in key-value pairs. Keys must be unique and immutable (eg. strings, numbers, or tuples), which values can be of any type.

```python
##Creating Dictionaries
empty_dict={}
print(type(empty_dict))
```

```python
empty_dict=dict()
empty_dict
```

```python
student={"name":"Subodh", "age":32, "grade":24}
print(student)
print(type(student))
```

```python
##Single Key is always used
student={"name":"Subodh", "age":32, "name":24}
print(student)
```

```python
##accessing Dictionary elements
student={"name":"Subodh", "age":32, "grade":'A'}
print(student)
```

```python
print(student['grade'])
print(student['age'])

##Acessing using get() method
print(student.get('grade'))
print(student.get('last_name'))
print(student.get('last_name',"Not available"))
```

```python
##Modifying Dictionary elements
##Dictionary are mutable, so you can add, update, or delete elements

print(student)
```

```python
student["age"]=33 #3update the value for the key
print(student)
student["address"]="India" ##added a new key and value
print(student)
```

```python
del student['grade'] ## delete key and value pair
print(student)
```

```python
##Dictionary methods
keys=student.keys() ##get all the keys
print(keys)
values=student.values() ##get all values
print(values)

items=student.items() ##get all key value pairs
print(items)
```

```python
##shallow copy
student_copy=student
print(student)
print(student_copy)
```

```python
student["name"]="Subodh2"
print(student)
print(student_copy)
```

```python
student_copy1=student.copy() #3shallow copy
print(student_copy1)
print(student)
```

```python
student["name"]="Subodh3"
print(student_copy1)
print(student)
```

```python
##Nested Dictionaries
students={
    "student1":{"name":"Subodh", "age":31},
    "student2":{"name":"Lata", "age":25},
}
print(students)
```

```python
##Access nested dictionaries elements
print(students["student2"]["name"])
print(students["student2"]["age"])
```

```python
##Iterating over nestetd dictionaries
for student_id, student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key, value in student_info.items():
        print(f"{key}:{value}")
```

```python
##Dictionary Comprehension
squares={x:x**2 for x in  range(5)}
print(squares)
```

```python
##Condional dictionaries comprehension
evens={x:x**2 for x in range(10) if  x%2==0}
print(evens)
```

```python
##Condional dictionaries comprehension
odds={x:x**2 for x in range(10) if  x%2!=0}
print(odds)
```

```python
##Practical Example
##use a dictionary to count he frequency  of elemnts in list
numbers=[1,2,2,3,3,3,4,4,4,4]
frequency={}

for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1
print(frequency)    
```

```python
## Merge 2 dictionaries into one
dict1={"a":1,"b":2}
dict2={"b":3,"c":4}
merged_dict = {**dict1, **dict2}
print(merged_dict)
```
